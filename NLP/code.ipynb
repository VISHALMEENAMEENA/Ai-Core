```python
import nltk 
from nltk.tokenize import sent_tokenize,word_tokenize,wordpunct_tokenize,TreebankWordTokenizer
corpus = 'Hello welcome to the NLP tutorials. Please  do watch the entire course! To become expert in NLP. hey .'

nltk.download("punkt_tab")
```

```python
print(sent_tokenize(corpus)) # list of sentences
print(word_tokenize(corpus))  # words, punctuation as separate tokens
print(wordpunct_tokenize(corpus)) # splits ALL punctuation, even apostrophes (Krish's becomes Krish, ', s)
print(TreebankWordTokenizer().tokenize(corpus))
```

Stemming

```python
from nltk.stem import PorterStemmer,RegexpStemmer,SnowballStemmer

w="history"

poter =PorterStemmer()
regex =RegexpStemmer('ing$|s$|e$|able$', min=4)
snow =SnowballStemmer('english')

# print(PorterStemmer().stem(w))
# print(RegexpStemmer('ing$|s$|e$|able$', min=4).stem(w))
# print(SnowballStemmer('english').stem(w))

for w in ['eating','eats','eaten','writing','programming','history','finally','finalized']:
    print(w,'---->',poter.stem(w),"  ",regex.stem(w)," ",snow.stem(w))
```

Lemmatization

```python
from nltk.stem import WordNetLemmatizer
nltk.download("wordnet")  # need to download it 


w="history"
lem=WordNetLemmatizer()

print(lem.lemmatize(w,pos='v'))
print(lem.lemmatize('eating',pos='n'))
print(lem.lemmatize('eating',pos='v'))
```

Stopwords

```python
from nltk.corpus  import stopwords
from nltk.tokenize import word_tokenize

# nltk.download('stopwords')

st=set(stopwords.words('english'))

sent="The students were sitting in the classroom, and they were listening to a lesson about language. Although the teacher explained the topic clearly, some of them were still unsure about how words are used in a sentence"
sentences=sent_tokenize(sent)
print(sentences)

for i in range(len(sentences)):
    words=word_tokenize(sentences[i])
    words=[w for w in words if w.lower() not in st]

    sentences[i]=' '.join(words)
print(sentences)
```

Named Entity Recognition (NER)

```python
# nltk.download('maxent_ne_chunker')
# nltk.download('words')
sentence = "On October 24, 2026, Sundar Pichai announced that Google would invest 10 billion dollars in New York City."

words=sentence.split()
tagged =nltk.pos_tag(words)

print(tagged)
tree=nltk.ne_chunk(tagged)
# tree.draw()  // to draw diagram
print(tree)
```
